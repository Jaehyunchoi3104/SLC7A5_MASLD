```python
import celltypist
from celltypist import models
# Load your AnnData
import scanpy as sc
```

```python
adata_scsp=sc.read("/home/jaehyunchoi/MASLD_collaboration/multiomics_analysis/adata_scsp01.h5ad")
```

```python
models.download_models(force_update = True)
```

```python
models.models_description()
```

```python
model = models.Model.load(model = 'Healthy_Human_Liver.pkl')
```

```python
# Run annotation
predictions = celltypist.annotate(adata_scsp, model=model, majority_voting=True)
```

```python
print(predictions)
print(predictions.predicted_labels.head())
```

```python
adata_scsp.obs['celltypist_labels'] = predictions.predicted_labels['majority_voting'].values
```

```python
print(adata_scsp.obs['celltypist_labels'].value_counts())
```
